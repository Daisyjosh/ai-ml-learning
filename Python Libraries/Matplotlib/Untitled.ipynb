```python
import matplotlib.pyplot as plt
```

```python
# types of plots / charts
# line plot
# pie chart
# histograms
# scatter plots
```

```python
#line plot
x = [1,2,3,4,5]
y = [6,7,8,9,10]
plt.plot(x,y) 
plt.show()
```

```python
x = [1,4,5,7,8]
y = [3,5,2,6,1]
plt.plot(x,y)
plt.title("Example plot")
plt.xlabel("x values")
plt.ylabel("y values")
```

```python
x = [1,2,3,4,5]
y1 = [3,5,4,6,7]
y2 = [5,6,4,3,2]
plt.plot(x,y1,label="first example")
plt.plot(x,y2,label="second example")
plt.title("Example: 2 plot")
plt.xlabel("x values")
plt.ylabel("y values")
plt.legend()
```

```python
# marker, line, color
x = [1,2,3,4,5]
y1 = [3,5,4,6,7]
y2 = [5,6,4,3,2]
y3 = [6,5,4,7,5]
plt.plot(x,y1,"o--c",label="first example")
plt.plot(x,y2,"<-r",label="second example")
plt.plot(x,y3,color="orange",marker="o",linestyle="-.",linewidth='5',label="third example")
plt.title("Example: 3 plot")
plt.xlabel("x values")
plt.ylabel("y values")
plt.legend()
```

```python
# additional information
x = [1,2,3,4,5]
y1 = [3,5,4,6,7]
y2 = [5,6,4,3,2]
plt.plot(x,y1,label="first example")
plt.plot(x,y2,label="second example")
plt.title("Example: 2 plot")
plt.xlabel("x values")
plt.ylabel("y values")
plt.legend()
print(plt.style.available)
# plt.style.use("ggplot")
plt.style.use("default")
plt.grid()
plt.show()
```

```python
# # # slight cartoonish feel style - xkcd
with plt.xkcd():
    x = [1,2,3,4,5]
    y1 = [3,5,4,6,7]
    y2 = [5,6,4,3,2]
    plt.plot(x,y1)
    plt.plot(x,y2)
    plt.title("Example: 2 plot")
    plt.xlabel("x values")
    plt.ylabel("y values")
```

```python
# tight_layout - help us from overlaping x and y values
cgpa = [8.81,8.4,8.6,8.7,9.32]
semester = ["first","second","third","fourth","fifth"]
plt.plot(semester,cgpa)
plt.title("Daisy's CGPA thoroughout the semesters")
plt.xlabel("Semesters")
plt.ylabel("CGPA")
plt.tight_layout()
plt.show()
plt.style.use("default")
plt.savefig("CGPA Graph.png") # to save the graph
```

```python
# plot methods
# line plot -> plt.plot(x,y)
plt.style.use("default")
# Vertical Bar Charts
plt.bar(x,2)
plt.show()

# Horizontal Bar Charts
plt.barh(y,2)
plt.show()

# Scatter plots
plt.scatter(x,y)
plt.show()

# pie chat
plt.pie(x)
plt.show()

# Histogram - similar to bar chats
plt.hist(x,4)
plt.show()

# Box plots
plt.boxplot(y)
plt.show()
```

```python
# Bar charts

plt.bar(semester,cgpa,color="purple")
plt.title("Daisy cgpa")
plt.xlabel("Semester")
plt.ylabel("cgpa")
plt.show()
```

```python
# detailed bar graph
# plt.text()
plt.style.use("default")
plt.bar(semester,cgpa,color="purple")
plt.title("Daisy cgpa")
plt.xlabel("Semester")
plt.ylabel("cgpa")
for i in range(len(cgpa)):
    plt.text(semester[i],cgpa[i]+0.2,str(cgpa[i]),ha="center",color="black")
# plt.ylim(0,max(cgpa)) change the axis limit
```

```python
# bar plot comparision
plt.style.use("default")
subjects = ["Maths", "Physics", "Chemistry", "Python", "DSA"]

internal_marks = [18, 20, 17, 19, 16]   # out of 20
external_marks = [72, 68, 70, 75, 74]   # out of 80

plt.bar(subjects,internal_marks)
plt.title("Internal Marks",color="Black")
plt.xlabel("Subject")
plt.ylabel("Internal Marks")
plt.show()

plt.bar(subjects,external_marks)
plt.title("External Marks",color="Black")
plt.xlabel("Subjects")
plt.ylabel("Exteranl Marks")
plt.show()

# clumpsy comparision
plt.bar(subjects,internal_marks,label="internal marks")
plt.bar(subjects,external_marks,label="external marks")
plt.title("Internal Marks vs External Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.legend()
plt.show()

# neat comparision
import numpy as np
x = np.arange(len(subjects))

width = 0.4
plt.bar(x-width/2,internal_marks,label="internal marks",width=width)
plt.bar(x+width/2,external_marks,label="external marks",width=width)
plt.title("Internal Marks vs External Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.legend()
plt.xticks(x,subjects) # to change the x value to respective values.
plt.show()
```
