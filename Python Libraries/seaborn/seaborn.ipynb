```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
print(type(tips))
print(tips.head())
```

```python
sns.relplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "smoker",
    size = "size",
    style = "smoker"
)
```

```python
sns.relplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    kind = "line"
)
```

```python
# scatter plot

sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "time"
    
)
```

```python
# line plot
sns.lineplot(
    data = tips,
    x = "total_bill",
    y = "tip"
)
```

```python
flights = sns.load_dataset("flights")
print(flights.head())
sns.lineplot(
    data = flights,
    x = "year",
    y = "passengers"
    
)
```

```python
# Categorical plots
# Bar plots
# Box plots

sns.barplot(
    data = tips,
    x = "day",
    y = "tip",
    hue = "sex"
)
```

```python
# Box plot

sns.boxplot(
    data = tips,
    x = "day",
    y = "tip"
)
```

```python
# Distribution plots
# Histogram

penguins = sns.load_dataset("penguins")
sns.histplot(
    data = penguins,
    x = "body_mass_g",
    bins = 10
)
```

```python
# Matrix plots
# HeatMaps - plot rectangular data as color encoded matrix

flights_pivot = flights.pivot(index="month",columns="year",values="passengers")
sns.heatmap(
    flights_pivot,
    cmap = "coolwarm"
)
```

```python

flights_pivot = flights.pivot(index="month",columns="year",values="passengers")
sns.heatmap(
    flights_pivot,
    cmap = "coolwarm",
    annot = True,
    fmt = "d"
    
)
```

```python
# Matplotlib with seaborn
fig, ax = plt.subplots()
sns.lineplot(
    data = tips,
    x = "day",
    y = "total_bill",
    hue = "sex",
    marker = "o",
    ax = ax,
    errorbar = None
)

ax.set_title("Avg bill value for each day")
ax.set_xlabel("Days")
ax.set_ylabel("Total Bill Amount")

fig.tight_layout()
```
